```python
# Import Libraries
import pandas as pd
import os

# Change to your path (the Data_Cleaning folder)
path = '/Users/nicolemorenogonzalez/Documents/STA-370/project-FMA/Data_Cleaning'

# Read in metadata (tracks.csv has a two-row header)
tracks = pd.read_csv(os.path.join(path, 'fma_metadata', 'tracks.csv'), index_col=0, header=[0, 1])
genres = tracks[('track', 'genre_top')]
splits = tracks[('set', 'split')] # training, validation or test
subsets = tracks[('set', 'subset')] # small, medium or large

# Keep fma_small tracks in our four genres
rows = []
for track_id in tracks.index:
    if subsets[track_id] != 'small' or genres[track_id] not in ['Rock', 'Folk', 'Pop', 'Electronic']:
        continue

    # Audio path, e.g. track 10 -> fma_small/000/000010.mp3
    file_name = str(track_id).zfill(6) + '.mp3'
    audio_path = os.path.join('fma_small', file_name[:3], file_name)

    # Skip broken files (a few are ~1 KB with no audio)
    if os.path.getsize(os.path.join(path, audio_path)) < 10000:
        continue

    rows.append([track_id, genres[track_id], splits[track_id], audio_path])

df = pd.DataFrame(rows, columns=['track_id', 'genre', 'split', 'audio_path'])

df.to_csv(os.path.join(path, 'cleaned_data', 'tracks_clean.csv'), index=False)
```
